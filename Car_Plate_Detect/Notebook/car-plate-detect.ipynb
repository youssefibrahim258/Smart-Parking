```python
from ultralytics import YOLO

dataset = '/kaggle/working/experiment-plates-1'
# Initialize the model
model = YOLO('yolo11m.pt')
# Train the model
results = model.train(
    data=f'{dataset}/data.yaml',
    epochs=30,           
    batch=8,              # Lower batch size to avoid overfitting and GPU OOM
    imgsz=640,            # Keep as is unless you know a better resolution for your task
    lr0=0.003,            # Lower initial LR to avoid divergence on small datasets
    lrf=0.001,             # Smaller decay for a more gradual LR reduction
    momentum=0.937,
    weight_decay=0.0005,
    patience=5,# Early stopping if no improvement
    cache=True,
# Cache data in memory to speed up training
    augment=True,         # Enable data augmentation (default in Ultralytics)


)
```
